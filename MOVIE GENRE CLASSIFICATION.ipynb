```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
import re

def read_dataset(file_path):
    data = []
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as file:
        for line in file:
            parts = line.strip().split(' ::: ')
            if len(parts) == 4:
                index, title, genre, plot = parts
                data.append([plot, genre])
    return pd.DataFrame(data, columns=['plot', 'genre'])

train_df = read_dataset('Movie_train_data.txt')
test_df = read_dataset('Movie_test_data.txt')

print("Training DataFrame:")
print(train_df.head())
print("Testing DataFrame:")
print(test_df.head())

assert not train_df.empty, "Training DataFrame is empty!"
assert not test_df.empty, "Testing DataFrame is empty!"

```

## Data Preprocessing

```python
train_df['plot'] = train_df['plot'].str.lower().apply(lambda x: re.sub(r'[^\w\s]', '', x))
test_df['plot'] = test_df['plot'].str.lower().apply(lambda x: re.sub(r'[^\w\s]', '', x))
```

## Feature Extraction

```python
tfidf = TfidfVectorizer(stop_words='english', max_features=5000)
X_train = tfidf.fit_transform(train_df['plot'])
X_test = tfidf.transform(test_df['plot'])
```

```python
y_train = train_df['genre']
y_test = test_df['genre']
```

## Model training and evaluation

```python
def train_and_evaluate_model(model, X_train, y_train, X_test, y_test):
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    print(f"Model: {model.__class__.__name__}")
    print("Accuracy:", accuracy_score(y_test, predictions))
    print(classification_report(y_test, predictions))
    print("-" * 60)

nb_model = MultinomialNB()
train_and_evaluate_model(nb_model, X_train, y_train, X_test, y_test)

lr_model = LogisticRegression(max_iter=1000)
train_and_evaluate_model(lr_model, X_train, y_train, X_test, y_test)
```
